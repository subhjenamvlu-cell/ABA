```python
import pandas as pd
df = pd.read_csv("StudentsPerformance.csv")
print(df.head())
```

```python
scores = df["math score"]
print("Mean :", scores.mean())
print("Median :", scores.median())
print("Mode :")
print(scores.mode())
print("Minimum :", scores.min())
print("Maximum :", scores.max())
print("Range :", scores.max() - scores.min())
print("Variance :", scores.var())
print("Standard Deviation :", scores.std())
print("Count :", scores.count())
print("Sum :", scores.sum())
print("\nDescriptive Statistics")
print(scores.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
scores = df["reading score"]
plt.figure(figsize=(8,5))
plt.hist(scores, bins=10, color="lightgreen", edgecolor="black")
plt.title("Histogram of Reading Scores")
plt.xlabel("Reading Score")
plt.ylabel("Frequency")
plt.show()
```

```python
scores = df["reading score"]
plt.figure(figsize=(8,3))
sns.boxplot(x=scores, color="orange")
plt.title("Box Plot of Reading Scores")
plt.xlabel("Reading Score")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Retail_sales.csv")
```

```python
# Count products in each category
count = df["Product Category"].value_counts()
print("Products in Each Category:")
print(count)
```

```python
# Sales by category
sales = df.groupby("Product Category")["Sales Revenue (USD)"].sum()
print("\nSales by Category:")
print(sales)
```

```python
# Highest sales category
print("\nCategory with Highest Sales:")
print(sales.idxmax())
```

```python
# Bar Chart
sales.plot(kind="bar")
plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Sales Revenue")
plt.show()
```

```python
store = df.groupby("Store ID")["Sales Revenue (USD)"].sum()

print("Total Sales of Each Store:")
print(store)

print("\nBest Performing Store:")
print(store.idxmax())

print("\nLeast Performing Store:")
print(store.idxmin())

# Horizontal Bar Chart
store.plot(kind="barh")
plt.title("Store Sales Revenue")
plt.xlabel("Sales Revenue")
plt.ylabel("Store ID")
plt.show()
```

```python
location = df.groupby("Store Location")["Sales Revenue (USD)"].sum()

print(location)

print("\nLocation with Highest Sales:")
print(location.idxmax())

# Bar Chart
location.plot(kind="bar")
plt.title("Sales by Store Location")
plt.xlabel("Store Location")
plt.ylabel("Sales Revenue")
plt.show()
```

```python
x = df["Units Sold"]

print("Mean =", x.mean())
print("Median =", x.median())
print("Mode =", x.mode()[0])
print("Minimum =", x.min())
print("Maximum =", x.max())
print("Range =", x.max() - x.min())
print("Variance =", x.var())
print("Standard Deviation =", x.std())
```

NetFlix

```python
# Read Dataset
df = pd.read_csv("NetFlix.csv")
```

```python
plt.boxplot(df["release_year"])
plt.title("Box Plot of Release Year")
plt.show()

print("Minimum =", df["release_year"].min())
print("Maximum =", df["release_year"].max())
print("Median =", df["release_year"].median())

Q1 = df["release_year"].quantile(0.25)
Q3 = df["release_year"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["release_year"] < lower) | (df["release_year"] > upper)]

print("Number of Outliers =", len(outliers))
```

```python
# Keep only Movies
movies = df[df["type"] == "Movie"].copy()

# Convert duration column to string before extracting numbers
movies["duration"] = movies["duration"].astype(str).str.extract(r"(\d+)").astype(float)

# Calculate statistics
print("Mean =", movies["duration"].mean())
print("Median =", movies["duration"].median())
print("Mode =", movies["duration"].mode()[0])
print("Variance =", movies["duration"].var())
print("Standard Deviation =", movies["duration"].std())
```
